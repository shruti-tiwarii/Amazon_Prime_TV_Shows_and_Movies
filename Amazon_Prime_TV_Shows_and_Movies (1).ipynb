## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
import ast

# ── 1. Parse list-encoded string columns ─────────────────────────────────────
def safe_parse_list(val):
    """Safely parse a stringified Python list (e.g. "['US', 'GB']") into a real list."""
    try:
        return ast.literal_eval(val) if isinstance(val, str) else []
    except (ValueError, SyntaxError):
        return []

titles_df['genres_list']   = titles_df['genres'].apply(safe_parse_list)
titles_df['countries_list'] = titles_df['production_countries'].apply(safe_parse_list)

# ── 2. Derive a primary genre and primary country for easier analysis ─────────
titles_df['primary_genre']   = titles_df['genres_list'].apply(lambda x: x[0].title() if x else 'Unknown')
titles_df['primary_country'] = titles_df['countries_list'].apply(lambda x: x[0] if x else 'Unknown')

# ── 3. Create a decade column ────────────────────────────────────────────────
titles_df['decade'] = (titles_df['release_year'] // 10 * 10).astype(str) + 's'

# ── 4. Fill missing age_certification as 'Not Rated' ─────────────────────────
titles_df['age_certification'] = titles_df['age_certification'].fillna('Not Rated')

# ── 5. Rename 'type' to 'show_type' for clarity (if column is named 'type') ──
if 'type' in titles_df.columns and 'show_type' not in titles_df.columns:
    titles_df.rename(columns={'type': 'show_type'}, inplace=True)

# ── 6. Impute missing scores with median (for visualization purposes) ─────────
titles_df['imdb_score_filled'] = titles_df['imdb_score'].fillna(titles_df['imdb_score'].median())
titles_df['tmdb_score_filled'] = titles_df['tmdb_score'].fillna(titles_df['tmdb_score'].median())

# ── 7. Explode genres for genre-level analysis ────────────────────────────────
genres_exploded = titles_df[['id', 'title', 'show_type', 'imdb_score', 'release_year']].copy()
genres_exploded['genre'] = titles_df['genres_list']
genres_exploded = genres_exploded.explode('genre').dropna(subset=['genre'])
genres_exploded['genre'] = genres_exploded['genre'].str.title()


# ── 8. Explode countries for country-level analysis ───────────────────────────
countries_exploded = titles_df[['id', 'title', 'show_type', 'imdb_score', 'release_year']].copy()
countries_exploded['country'] = titles_df['countries_list']
countries_exploded = countries_exploded.explode('country').dropna(subset=['country'])

print("✅ Data wrangling complete.")
print(f"   genres_exploded rows   : {len(genres_exploded):,}")
print(f"   countries_exploded rows: {len(countries_exploded):,}")
display(titles_df[['title','show_type','primary_genre','primary_country','decade','age_certification']].head())
````

### What all manipulations have you done and insights you found?

1. **Parsed list columns** (`genres`, `production_countries`): These were stored as stringified Python lists. Using `ast.literal_eval` converted them into actual Python lists, enabling genre and country explosion.
2. **Primary genre / country** extraction: Took the first element from each list for simplified categorical analysis.
3. **Decade column**: Binned `release_year` into decades to identify long-term trends in content production.
4. **Age certification imputation**: Replaced NaN with `'Not Rated'` to retain all rows in categorical analysis.
5. **Score imputation with median**: Created `imdb_score_filled` and `tmdb_score_filled` for visualizations that require no missing values (e.g., scatter plots).
6. **Exploded genre and country DataFrames**: Created long-format tables for accurate multi-genre and multi-country counting.
**Key Insight**: The dataset is predominantly movies (80%), US-produced content dominates (50%+), and drama/comedy are the most frequent genres.

````python
# Chart - 1 visualization code
# Movies vs Shows split (Pie Chart)
type_counts = titles_df['show_type'].value_counts()

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Pie chart
colors = ['#00A8E1', '#FF9900']
wedges, texts, autotexts = axes[0].pie(
    type_counts,
    labels=type_counts.index,
    autopct='%1.1f%%',
    colors=colors,
    startangle=90,
    wedgeprops={'edgecolor': 'white', 'linewidth': 2},
    textprops={'fontsize': 13}
)
for at in autotexts:
    at.set_fontsize(13)
    at.set_fontweight('bold')
axes[0].set_title('Content Type Distribution', fontsize=14, fontweight='bold')

# Bar chart
bars = axes[1].bar(type_counts.index, type_counts.values, color=colors, edgecolor='white', linewidth=1.5, width=0.5)
for bar, val in zip(bars, type_counts.values):
    axes[1].text(bar.get_x() + bar.get_width()/2, bar.get_height() + 50, f'{val:,}',
                 ha='center', va='bottom', fontsize=12, fontweight='bold')
axes[1].set_title('Count of Movies vs Shows', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Content Type')
axes[1].set_ylabel('Count')
axes[1].set_ylim(0, type_counts.max() * 1.15)

plt.suptitle('Amazon Prime — Movies vs TV Shows', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
````

A horizontal bar chart sorted by score value provides a clear ranking of genres by quality perception, with the vertical mean line as a reference to identify above- and below-average genres at a glance.

##### 2. What is/are the insight(s) found from the chart?

Documentary and History genres receive the highest average IMDB scores (~7.5+), while Horror, Comedy, and Thriller tend to score lower on average. This reflects audience voting patterns where informational content earns respect, while mass-market genres have higher variance.
